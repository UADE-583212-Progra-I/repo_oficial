## Bloque 1 - Validación

### Validar dígitos

```python
# Definir un patron y un str, evaluar con de fullmatch
patron = r"[0-9]+" # completar
cadena = "1t" # completar

print(bool(re.fullmatch(patron, cadena)))
```

### Validar nombres

```python
# Definir un patron y un str y evaluar con de fullmatch
patron = r"[a-zA-Z ]+" # completar
cadena = "mariano oscar" # completar

print(bool(re.fullmatch(patron, cadena)))
```

### Validar emails

```python
# Definir un patron y un str y evaluar con de fullmatch
patron = r"[a-zA-Z]+@\w+[\.\w]+" # completar
cadena = "estudiante@uade.edu" # completar

print(bool(re.fullmatch(patron, cadena)))
```

```python
def buscar_tema_por_nombre(nombre:str, temas:list):
    return [t for t in temas if re.search(nombre, t["nombre"], re.IGNORECASE)] # Completar

print(buscar_tema_por_nombre("love$", temas)) # reemplazar y evaluar
```
