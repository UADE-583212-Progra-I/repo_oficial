## Parte 1 — Lista simple de ratings

Empezamos con lo más básico: una lista de números que representa las calificaciones que los usuarios dieron a los temas.


````python
# Punto de partida: una lista plana de ratings
ratings_simples = [5, 3, 4, 1, 5, 2, 4, 3, 5, 1, 4, 2]

print("Ratings registrados:", ratings_simples)
print("Total de ratings:  ", len(ratings_simples))
````

### 1.1 Lista por comprensión — estructura básica

La estructura es:

```python
[expresion  for elemento in lista]                    # transformación
[expresion  for elemento in lista  if condicion]      # transformación + filtro
```

Antes de las listas por comprensión, usábamos `for` + `append`:


````python
# Forma tradicional: ratings altos (>= 4) con for + append
ratings_altos_tradicional = []
for r in ratings_simples:
    if r >= 4:
        ratings_altos_tradicional.append(r)

print("Forma tradicional:", ratings_altos_tradicional)
````

````python
ratings_simples
````

````python
# Lista por comprensión: exactamente lo mismo en una línea
ratings_altos = [r for r in ratings_simples if r >=4]

print("Lista por comprensión:", ratings_altos)
````

````python
# Transformación: convertir cada rating a porcentaje sobre 5
# rating 4 → 80%,  rating 5 → 100%,  etc.
porcentajes = [r * 100 // 5 for r in ratings_simples]

print("Ratings:     ", ratings_simples)
print("Porcentajes: ", porcentajes)
````

````python
# Transformación + filtro combinados
# Solo los ratings >= 4, convertidos a porcentaje
porcentajes_altos = [r * 100 // 5 for r in ratings_simples if r >= 4]

print("Porcentajes de ratings altos:", porcentajes_altos)
````

### 1.2 `lambda` — función anónima de una línea

Una `lambda` es una función sin nombre, definida en el momento:

```python
# función normal
def duplicar(x):    
    return x * 2

# lambda equivalente
duplicar = lambda x: x * 2
```

Se usa cuando la función es tan simple que no vale la pena darle nombre ni usar `def`.


````python
# Función normal vs. lambda — hacen lo mismo
def es_rating_alto(r:int):
    return r >= 4 # true/false

es_alto = lambda r: r >= 4

# Probar con un valor
print("Con def:    ", es_rating_alto(3), es_rating_alto(5))
print("Con lambda: ", es_alto(3),        es_alto(5))
````

### 1.3 `map` — transformar cada elemento

````python
# map(función, lista) → aplica la función a cada elemento
# Siempre hay que convertir con list() para ver el resultado

# Convertir cada rating a porcentaje con map + lambda
# porcentaje = lambda r: r * 100 // 5
porcentajes_map = list(map(lambda r: r * 100 // 5, ratings_simples))

print("Ratings:     ", ratings_simples)
print("map+lambda:  ", porcentajes_map)

# Es equivalente a la lista por comprensión:
porcentajes_comp = [r * 100 // 5 for r in ratings_simples]
print("comprensión: ", porcentajes_comp)
print("¿Son iguales?", porcentajes_map == porcentajes_comp)
````

### 1.4 `filter` — conservar solo los que cumplen una condición

````python
# filter(función, lista) → conserva los elementos donde la función devuelve True
ratings_altos_filter = list(filter(lambda r: r >= 4, ratings_simples))

print("Ratings originales: ", ratings_simples)
print("filter >= 4:        ", ratings_altos_filter)
````

### 1.5 `sorted` — ordenar por criterio

````python
# sorted con lambda: ordenar de mayor a menor
ordenados = sorted(ratings_simples, key=lambda r: r, reverse=True)
print("Mayor a menor:", ordenados)

# Para listas simples también funciona sin lambda:
print("Sin lambda:   ", sorted(ratings_simples, reverse=True))
# La lambda es más útil con listas de diccionarios (lo vemos en la Parte 2)
````

### 1.6 `reduce` — acumular en un único valor

````python
from functools import reduce

# reduce(función, lista) aplica la función acumulando de izquierda a derecha:
# paso 1: f(ratings[0], ratings[1])
# paso 2: f(resultado_anterior, ratings[2])
# ...y así hasta el final

# Suma total con reduce
suma = reduce(lambda acumulador, r: acumulador + r, ratings_simples)
print("Suma total:", suma)

# Promedio
promedio = suma / len(ratings_simples)
print(f"Promedio:  {promedio:.2f}")

# Rating máximo con reduce (sin usar max())
maximo = reduce(lambda acc, r: acc if acc > r else r, ratings_simples)
print("Máximo:   ", maximo)
````

## Parte 2 — Lista de diccionarios: ratings con contexto

La lista de números es útil para entender los conceptos, pero en el sistema real cada rating es un diccionario con `id_usuario`, `id_tema` y `rating`. Vamos a trabajar con esa estructura.


````python
# Datos del sistema — igual que en la actividad integradora
TEMAS = [
    {"id_tema": 1, "tema": "Dai Dai",        "autor": "Shakira"},
    {"id_tema": 2, "tema": "Dynamite",        "autor": "BTS"},
    {"id_tema": 3, "tema": "DTMF",            "autor": "Bad Bunny"},
    {"id_tema": 4, "tema": "Dont Start Now",  "autor": "Dua Lipa"},
    {"id_tema": 5, "tema": "Positions",       "autor": "Ariana Grande"},
]

USUARIOS = [
    {"id_usuario": 1, "nombre": "Lorenzo",  "apellido": "Pérez"},
    {"id_usuario": 2, "nombre": "Mariano",  "apellido": "García"},
    {"id_usuario": 3, "nombre": "Ignacio",  "apellido": "López"},
    {"id_usuario": 4, "nombre": "Agustin",  "apellido": "Martínez"},
    {"id_usuario": 5, "nombre": "Tatiana",  "apellido": "Rodríguez"},
]

RATINGS = [
    {"id_rating": 1,  "id_usuario": 1, "id_tema": 1, "rating": 5},
    {"id_rating": 2,  "id_usuario": 1, "id_tema": 2, "rating": 3},
    {"id_rating": 3,  "id_usuario": 1, "id_tema": 3, "rating": 4},
    {"id_rating": 4,  "id_usuario": 2, "id_tema": 1, "rating": 4},
    {"id_rating": 5,  "id_usuario": 2, "id_tema": 3, "rating": 5},
    {"id_rating": 6,  "id_usuario": 2, "id_tema": 5, "rating": 2},
    {"id_rating": 7,  "id_usuario": 3, "id_tema": 2, "rating": 5},
    {"id_rating": 8,  "id_usuario": 3, "id_tema": 4, "rating": 4},
    {"id_rating": 9,  "id_usuario": 4, "id_tema": 1, "rating": 3},
    {"id_rating": 10, "id_usuario": 4, "id_tema": 5, "rating": 5},
    {"id_rating": 11, "id_usuario": 5, "id_tema": 3, "rating": 4},
    {"id_rating": 12, "id_usuario": 5, "id_tema": 4, "rating": 1},
]

print(f"Temas:    {len(TEMAS)}")
print(f"Usuarios: {len(USUARIOS)}")
print(f"Ratings:  {len(RATINGS)}")
````

### 2.1 Extraer una columna — lista por comprensión sobre diccionarios

````python
# Extraer solo los valores de rating (equivalente a una columna en Pandas)
valores = [r["rating"] for r in RATINGS]
print("Valores de rating:", valores)

# Extraer los id_tema de todos los ratings
temas_calificados = [r["id_tema"] for r in RATINGS]
print("Temas calificados:", temas_calificados)

# Temas únicos calificados (con set)
temas_unicos = list(set(temas_calificados))
print("Temas únicos:     ", sorted(temas_unicos))
````

### 2.2 Filtrar ratings por condición

````python
# Ratings >= 4 (positivos)
ratings_positivos = [r for r in RATINGS if r["rating"] >= 4]
print(f"Ratings positivos (>= 4): {len(ratings_positivos)} de {len(RATINGS)}")
for r in ratings_positivos:
    print(f"  rating {r['id_rating']}: usuario {r['id_usuario']} → tema {r['id_tema']} = {r['rating']}")
````

````python
# Ratings de un usuario específico
id_usuario_buscar = 1
ratings_usuario1 = [r for r in RATINGS if r["id_usuario"] == id_usuario_buscar]

print(f"Ratings del usuario {id_usuario_buscar}:")
for r in ratings_usuario1:
    print(f"  tema {r['id_tema']}: {r['rating']}/5")
````

````python
# Ratings de un tema específico (con filter + lambda)
id_tema_buscar = 3
ratings_tema3 = list(filter(lambda r: r["id_tema"] == id_tema_buscar, RATINGS))

print(f"Ratings del tema {id_tema_buscar} (DTMF — Bad Bunny):")
for r in ratings_tema3:
    print(f"  usuario {r['id_usuario']}: {r['rating']}/5")
````

### 2.3 Transformar: agregar campos calculados con `map`

````python
# Agregar campo "positivo" (True si rating >= 4) usando map + lambda
# La lambda construye un nuevo diccionario con el campo extra

ratings_con_flag = list(map(
    lambda r: {**r, "positivo": r["rating"] >= 4},
    RATINGS
))

print("Ratings con campo 'positivo':")
for r in ratings_con_flag[:5]:
    print(f"  {r}")
````

````python
# Agregar campo "porcentaje" = rating / 5 * 100
ratings_con_pct = list(map(
    lambda r: {**r, "porcentaje": r["rating"] * 100 // 5},
    RATINGS
))

print("Primeros 5 ratings con porcentaje:")
for r in ratings_con_pct[:5]:
    print(f"  usuario {r['id_usuario']} → tema {r['id_tema']}: {r['rating']}/5 ({r['porcentaje']}%)")
````

### 2.4 Ordenar con `sorted` + `lambda`

````python
# Ordenar ratings de mayor a menor calificación
ratings_ordenados = sorted(RATINGS, key=lambda r: r["rating"], reverse=True)

print("Top 5 ratings más altos:")
for r in ratings_ordenados[:5]:
    print(f"  usuario {r['id_usuario']} → tema {r['id_tema']}: {r['rating']}/5")
````

````python
# Ordenar usuarios alfabéticamente por nombre
usuarios_ordenados = sorted(USUARIOS, key=lambda u: u["nombre"])
print("Usuarios ordenados alfabéticamente:")
for u in usuarios_ordenados:
    print(f"  {u['nombre']} {u['apellido']}")
````

## Parte 3 — Estadísticas por tema

Ahora combinamos todo para calcular estadísticas reales del sistema: promedio de ratings por tema, tema mejor calificado, etc. Cada función usa las herramientas que acabamos de ver.


### 3.1 Promedio de ratings por tema

````python
from functools import reduce

def ratings_de_tema(id_tema):
    """Retorna la lista de valores de rating para un tema dado."""
    return [r["rating"] for r in RATINGS if r["id_tema"] == id_tema]

def promedio_tema(id_tema):
    """Calcula el rating promedio de un tema. Retorna 0 si no tiene ratings."""
    valores = ratings_de_tema(id_tema)
    if not valores:
        return 0
    suma = reduce(lambda acc, r: acc + r, valores)
    return round(suma / len(valores), 2)

# Mostrar promedio para cada tema
print("Rating promedio por tema:")
print(f"{'ID':<5} {'Tema':<22} {'Autor':<18} {'Ratings':<9} {'Promedio'}")
print("-" * 65)
for t in TEMAS:
    valores = ratings_de_tema(t["id_tema"])
    prom    = promedio_tema(t["id_tema"])
    print(f"  {t['id_tema']:<3} {t['tema']:<22} {t['autor']:<18} {len(valores):<9} {prom}")
````

### 3.2 Tema mejor y peor calificado

````python
# Construir lista de (tema, promedio) para todos los temas con al menos 1 rating
temas_con_prom = [
    {"tema": t["tema"], "autor": t["autor"], "promedio": promedio_tema(t["id_tema"])}
    for t in TEMAS
    if len(ratings_de_tema(t["id_tema"])) > 0
]

# Ordenar por promedio
temas_ordenados = sorted(temas_con_prom, key=lambda t: t["promedio"], reverse=True)

print("Ranking de temas por rating promedio:")
for i, t in enumerate(temas_ordenados, 1):
    print(f"  {i}. {t['tema']} — {t['autor']}: {t['promedio']}/5")

mejor = temas_ordenados[0]
peor  = temas_ordenados[-1]
print()
print(f"Mejor calificado: {mejor['tema']} ({mejor['promedio']})")
print(f"Peor calificado:  {peor['tema']} ({peor['promedio']})")
````

### 3.3 Perfil de cada usuario

````python
def perfil_usuario(id_usuario):
    """Retorna estadísticas de un usuario: cantidad de ratings, promedio y temas calificados."""
    mis_ratings = [r for r in RATINGS if r["id_usuario"] == id_usuario]

    if not mis_ratings:
        return {"cantidad": 0, "promedio": 0, "temas": []}

    valores   = [r["rating"] for r in mis_ratings]
    suma      = reduce(lambda acc, v: acc + v, valores)
    promedio  = round(suma / len(valores), 2)
    ids_temas = [r["id_tema"] for r in mis_ratings]

    return {
        "cantidad": len(mis_ratings),
        "promedio": promedio,
        "temas":    ids_temas,
    }

print("Perfil de cada usuario:")
print(f"{'Nombre':<12} {'Ratings':<9} {'Promedio':<10} {'Temas calificados'}")
print("-" * 55)
for u in USUARIOS:
    p = perfil_usuario(u["id_usuario"])
    print(f"  {u['nombre']:<10} {p['cantidad']:<9} {p['promedio']:<10} {p['temas']}")
````

### 3.4 Temas que un usuario NO calificó — candidatos a recomendar

````python
def temas_sin_calificar(id_usuario):
    """Retorna los temas que el usuario todavía no calificó."""
    # IDs de temas que SÍ calificó
    ids_calificados = {r["id_tema"] for r in RATINGS if r["id_usuario"] == id_usuario}

    # Temas que NO están en ese conjunto
    pendientes = [t for t in TEMAS if t["id_tema"] not in ids_calificados]

    return pendientes

# Mostrar para cada usuario
print("Temas pendientes por usuario (candidatos a recomendar):")
for u in USUARIOS:
    pendientes = temas_sin_calificar(u["id_usuario"])
    nombres = [t["tema"] for t in pendientes]
    print(f"  {u['nombre']:<10}: {nombres}")
````

## Parte 4 — Construir la matriz de ratings

En la actividad integradora construimos la matriz con `for` anidados. Ahora la construimos usando todo lo que aprendimos — el resultado es idéntico, pero el código es más compacto.


````python
# Versión con for anidados (como en la actividad integradora)
def construir_matriz_for(usuarios, temas, ratings):
    indice = {}
    for r in ratings:
        indice[(r["id_usuario"], r["id_tema"])] = r["rating"]

    matriz = []
    for u in usuarios:
        fila = []
        for t in temas:
            fila.append(indice.get((u["id_usuario"], t["id_tema"]), 0))
        matriz.append(fila)
    return matriz

# Versión con listas por comprensión
def construir_matriz_comprension(usuarios, temas, ratings):
    indice = {(r["id_usuario"], r["id_tema"]): r["rating"] for r in ratings}

    return [
        [indice.get((u["id_usuario"], t["id_tema"]), 0) for t in temas]
        for u in usuarios
    ]

# Verificar que producen el mismo resultado
matriz_for  = construir_matriz_for(USUARIOS, TEMAS, RATINGS)
matriz_comp = construir_matriz_comprension(USUARIOS, TEMAS, RATINGS)

print("¿Matrices iguales?", matriz_for == matriz_comp)
````

````python
# Mostrar la matriz
def mostrar_matriz(matriz, usuarios, temas):
    # Encabezado de columnas (temas)
    print(f"{'':12}", end="")
    for t in temas:
        print(f"{t['tema'][:9]:>10}", end="")
    print()
    print("  " + "-" * (10 + 10 * len(temas)))

    # Filas (usuarios)
    for i, u in enumerate(usuarios):
        print(f"  {u['nombre']:<10}", end="")
        for v in matriz[i]:
            celda = str(v) if v > 0 else "·"
            print(f"{celda:>10}", end="")
        print()

print("Matriz de ratings (· = sin calificar):")
print()
mostrar_matriz(construir_matriz_comprension(USUARIOS, TEMAS, RATINGS), USUARIOS, TEMAS)
````

````python
# Extra: promedio por columna (por tema) usando comprensión
# Cada columna es el índice j de la matriz

matriz = construir_matriz_comprension(USUARIOS, TEMAS, RATINGS)

print("Promedio de ratings por tema (excluyendo ceros):")
for j, t in enumerate(TEMAS):
    columna         = [matriz[i][j] for i in range(len(USUARIOS))]
    valores_reales  = [v for v in columna if v > 0]
    if valores_reales:
        prom = reduce(lambda acc, v: acc + v, valores_reales) / len(valores_reales)
        print(f"  {t['tema']:<22}: {prom:.2f}  (de {len(valores_reales)} ratings)")
    else:
        print(f"  {t['tema']:<22}: sin ratings")
````

## 📋 Resumen

| Herramienta | Estructura | Cuándo usarla |
|---|---|---|
| Lista por comprensión | `[expr for x in lista if cond]` | Transformar o filtrar en una línea |
| `lambda` | `lambda x: expresion` | Función simple sin necesidad de `def` |
| `map` | `list(map(lambda x: ..., lista))` | Transformar cada elemento |
| `filter` | `list(filter(lambda x: ..., lista))` | Filtrar por condición |
| `sorted` | `sorted(lista, key=lambda x: ...)` | Ordenar por criterio |
| `reduce` | `reduce(lambda acc, x: ..., lista)` | Acumular en un único valor |

### ¿Lista por comprensión o `map`/`filter`?

Ambas hacen lo mismo. La preferencia en Python es:
- **Lista por comprensión** → más legible para transformaciones y filtros simples
- **`map`/`filter`** → útiles cuando ya tenés la función definida o cuando se encadenan operaciones
