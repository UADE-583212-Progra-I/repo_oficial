# 1. Diccionarios

Un **diccionario** es una estructura de datos que almacena pares **clave → valor**.

```
diccionario = {
    "clave_1": valor_1,
    "clave_2": valor_2,
}
```

Características principales:
- Los valores se acceden por **nombre** (clave), no por posición
- Las claves son **únicas** — no puede haber dos iguales
- Son **mutables** — se pueden agregar, modificar y eliminar elementos
- Desde Python 3.7+, **conservan el orden de inserción**

> 💡 También se los conoce como *arreglos asociativos*, *tablas hash* o *mapas*.


## 1.1 Creación

````python
# Forma 1: llaves literales
artista = {
    "precio":      1,
    "nombre_artista":  "Banksy",
    "nacionalidad":    "Inglaterra"
}
print(artista)
````

````python
# Forma 2: función dict()
artista2 = dict(id_artista=2, nombre_artista="Kusama", nacionalidad="Japon")
print(artista2)
````

````python
# Forma 3: fromkeys — crear diccionario con claves fijas y valor por defecto
campos = ["id_artista", "nombre_artista", "nacionalidad"]
artista_vacio = dict.fromkeys(campos, None)
print(artista_vacio)
````

## 1.2 Acceso a elementos

````python
artista = {"id_artista": 1, "nombre_artista": "Banksy", "nacionalidad": "Inglaterra"}

# Acceso por clave directa
print(artista["nombre_artista"])          # → Banksy

# Acceso seguro con get() — retorna None si no existe (no lanza error)
print(artista.get("nacionalidad"))        # → Inglaterra
print(artista.get("premio"))              # → None (no lanza KeyError)
print(artista.get("premio", "no existe")) # → No existe (valor por defecto)
````

## 1.3 Iterar diccionarios

````python
artista = {"id_artista": 1, "nombre_artista": "Banksy", "nacionalidad": "Inglaterra"}

# Solo claves
print("Claves:"); print(list(artista.keys()))

# Solo valores
print("Valores:"); print(list(artista.values()))

# Pares clave-valor (items)
print("Items:")
for clave, valor in artista.items():
    print(f"  {clave}: {valor}")
````

## 1.4 Modificar, agregar y eliminar

````python
artista = {"id_artista": 1, "nombre_artista": "Banksy",
           "nacionalidad": "Inglaterra"}

# Agregar un campo nuevo
artista["estilo"] = "Street Art"
print("Después de agregar estilo:", artista["estilo"])

# Modificar un valor existente
artista["nacionalidad"] = "Gran Bretaña"
print("Nacionalidad actualizada:", artista["nacionalidad"])

# Eliminar con del (lanza KeyError si no existe)
del artista["estilo"]
print("Después de del:", "estilo" in artista)

# Eliminar con pop() — retorna el valor eliminado
nac_eliminada = artista.pop("nacionalidad", "No existía")
print("Nacionalidad eliminada:", nac_eliminada)
````

````python
# update() — fusionar otro diccionario
artista = {"id_artista": 1, "nombre_artista": "Banksy", "nacionalidad": "Inglaterra"}
extra = {"nacionalidad": "Gran Bretaña", "estilo": "Street Art"}

artista.update(extra)
print(artista)   # nacionalidad actualizada, estilo agregado
````

## 1.5 Lista de diccionarios — la tabla en código

````python
# Cada diccionario es un registro (fila)
# La lista de diccionarios es la tabla

artistas = [
    {"id_artista": 1, "nombre_artista": "Banksy",   "nacionalidad": "Inglaterra"},
    {"id_artista": 2, "nombre_artista": "Kusama",   "nacionalidad": "Japon"},
    {"id_artista": 3, "nombre_artista": "Koons",    "nacionalidad": "EEUU"},
    {"id_artista": 4, "nombre_artista": "Hirst",    "nacionalidad": "Inglaterra"},
    {"id_artista": 5, "nombre_artista": "Murakami", "nacionalidad": "Japon"},
]

# Mostrar todos los artistas
for a in artistas:
    print(f"[{a['id_artista']}] {a['nombre_artista']} — {a['nacionalidad']}")
````

````python
# Filtrar por nacionalidad
ingleses = []
for a in artistas:
    if a["nacionalidad"] == "Inglaterra":
        ingleses.append(a)
print("Artistas de Inglaterra:", [a["nombre_artista"] for a in ingleses])
````

# 2. Tuplas

Una **tupla** es una secuencia ordenada e **inmutable** de elementos.

```
tupla = (elemento_0, elemento_1, elemento_2)
```

Características:
- **Inmutable**: una vez creada, no se puede modificar
- **Ordenada**: los elementos tienen posición (índice)
- **Permite duplicados**
- Se puede acceder por índice igual que las listas
- Son más eficientes en memoria que las listas

> 💡 Usá tuplas cuando el dato no debe cambiar: coordenadas, registros históricos, retorno múltiple de funciones.


## 2.1 Creación

````python
# Tupla simple
punto = (32.73, -117.14)   # (latitud, longitud)
print(type(punto), punto)

# Tupla de un solo elemento — requiere la coma
solo_uno = (42,)
print(type(solo_uno), solo_uno)

# Sin paréntesis también es tupla (packing)
colores_rgb = 255, 128, 0
print(type(colores_rgb), colores_rgb)
````

## 2.2 Acceso e inmutabilidad

````python
# Tupla que representa un registro de visita a una exposición
# (id_visita, id_artista, id_obra, fecha)
visita_record = (1, 3, 7, "2024-06-01")

print("id_visita:  ", visita_record[0])
print("id_artista: ", visita_record[1])
print("fecha:      ", visita_record[-1])

# Slicing (igual que listas)
print("Primeros dos:", visita_record[:2])

# Inmutabilidad — esto genera TypeError
try:
    visita_record[0] = 99
except TypeError as e:
    print(f"Error: {e}")
````

## 2.3 Unpacking — desempaquetar tuplas

````python
# Unpacking: asignar elementos de la tupla a variables con nombre
id_visita, id_artista, id_obra, fecha = (1, 3, 7, "2024-06-01")
print(f"Visita #{id_visita}: artista {id_artista} expuso obra {id_obra} el {fecha}")

# Muy útil en funciones que retornan múltiples valores
def stats_obras(obras):
    """Retorna (mínimo, máximo, promedio) de precios de obras."""
    precios = [o["precio"] for o in obras]
    return min(precios), max(precios), round(sum(precios)/len(precios), 2)

obras_ejemplo = [
    {"id_obra": 1, "titulo": "Balloon Dog",   "precio": 58_405_000},
    {"id_obra": 2, "titulo": "For the Love",  "precio": 100_000_000},
    {"id_obra": 3, "titulo": "Infinity Room", "precio": 8_000_000},
]

minimo, maximo, promedio = stats_obras(obras_ejemplo)
print(f"min={minimo:,}, max={maximo:,}, promedio={promedio:,.0f}")
````

# 3. Conjuntos (sets)

Un **conjunto** es una colección **sin duplicados** y **sin orden**.

```
conjunto = {elemento_1, elemento_2, elemento_3}
conjunto = set([lista])
```

Características:
- **Sin duplicados**: agregar un elemento ya existente no tiene efecto
- **Sin orden**: no se puede acceder por índice
- **Mutable**: se pueden agregar y quitar elementos
- Soportan operaciones de conjuntos: **unión**, **intersección**, **diferencia**

> 💡 Usá conjuntos cuando necesitás saber si algo ya existe o eliminar duplicados rápidamente.


## 3.1 Creación

````python
# Desde una lista — elimina duplicados automáticamente
nacionalidades_raw = ["Inglaterra", "Japon", "EEUU", "Japon", "Inglaterra", "Francia", "EEUU"]
nacionalidades = set(nacionalidades_raw)
print("Nacionalidades únicas:", nacionalidades)
print("Tipo:", type(nacionalidades))

# Literal
favoritos = {"Banksy", "Kusama", "Murakami"}
print(favoritos)

# ATENCIÓN: {} sin nada crea un diccionario vacío, no un set
vacio_dict = {}
vacio_set  = set()
print(type(vacio_dict), type(vacio_set))
````

## 3.2 Operaciones de conjunto

````python
# Artistas que participaron en la Exposición A
expA = {"Banksy", "Kusama", "Koons"}

# Artistas que participaron en la Exposición B
expB = {"Kusama", "Koons", "Hirst", "Murakami"}

# Unión — participaron en al menos una exposición
print("Unión:         ", expA | expB)

# Intersección — participaron en ambas exposiciones
print("Intersección:  ", expA & expB)

# Diferencia — solo en Exposición A (no en B)
print("Solo en Exp A: ", expA - expB)

# Diferencia simétrica — en una pero no en las dos
print("Diferencia sim:", expA ^ expB)
````

## 3.3 Operaciones de membership y mutación

````python
artistas_registrados = {"Banksy", "Kusama", "Koons", "Hirst"}

# Verificar pertenencia (más rápido que en listas)
print("¿'Kusama' está registrada?",  "Kusama"   in artistas_registrados)
print("¿'Monet' está registrado?",   "Monet"    in artistas_registrados)

# Agregar un artista
artistas_registrados.add("Murakami")
print("Después de add:", artistas_registrados)

# Agregar varios
artistas_registrados.update(["Basquiat", "Warhol"])
print("Después de update:", artistas_registrados)

# Quitar un elemento — discard no lanza error si no existe
artistas_registrados.discard("Warhol")
print("Después de discard:", artistas_registrados)
````

# 4. ¿Cuándo usar cada estructura?

| Pregunta | Estructura | Razón |
|---|---|---|
| ¿Necesito acceder por nombre? | **Diccionario** | clave → valor |
| ¿El dato no debe cambiar nunca? | **Tupla** | inmutable |
| ¿Necesito retornar varios valores? | **Tupla** | unpacking |
| ¿Necesito eliminar duplicados? | **Conjunto** | sin repetidos |
| ¿Necesito saber si algo ya existe (rápido)? | **Conjunto** | O(1) en membership |
| ¿Necesito orden y posición? | **Lista** | acceso por índice |

En nuestro sistema vamos a combinar las tres:

```python
# Artista: diccionario (atributos con nombre)
artista = {"id_artista": 1, "nombre_artista": "Banksy", "nacionalidad": "Inglaterra"}

# Visita a exposición: tupla (registro histórico inmutable)
visita_tuple = (id_visita, id_artista, id_obra, fecha)

# Nacionalidades únicas: conjunto (sin duplicados, búsqueda rápida)
nacionalidades_disponibles = {"Inglaterra", "Japon", "EEUU", "Francia"}
```

## 📋 Resumen de métodos

### Diccionarios
| Método | Descripción |
|---|---|
| `d[key]` | Acceso por clave (KeyError si no existe) |
| `d.get(key, default)` | Acceso seguro |
| `d.keys()` / `.values()` / `.items()` | Iterar |
| `d.update(otro)` | Fusionar / actualizar |
| `d.pop(key)` | Eliminar y retornar valor |
| `d.setdefault(key, val)` | Acceder o insertar si no existe |
| `dict.fromkeys(iter, val)` | Crear con claves fijas |

### Tuplas
| Operación | Descripción |
|---|---|
| `t[i]` | Acceso por índice |
| `a, b = t` | Unpacking |
| `len(t)` | Longitud |
| `x in t` | Membership |

### Conjuntos
| Operación | Descripción |
|---|---|
| `s.add(x)` | Agregar un elemento |
| `s.update(iter)` | Agregar varios |
| `s.discard(x)` | Quitar (sin error) |
| `a \| b` | Unión |
| `a & b` | Intersección |
| `a - b` | Diferencia |
| `a ^ b` | Diferencia simétrica |
| `x in s` | Membership (O(1)) |
